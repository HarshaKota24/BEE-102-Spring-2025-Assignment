## Markov Transition Matrix from DNA Sequence

## Input DNA Sequence

```python
sequence=input("Input sequence")
```

## Initialize Transition Count Matrix

```python
nucs = ['A', 'C', 'G', 'T']
count_matrix = {n1: {n2: 0 for n2 in nucs} for n1 in nucs}
```

## Count Transitions in Sequence
We loop over each adjacent nucleotide pair and update the matrix.

```python
for i in range(len(sequence) - 1):
    curr = sequence[i]
    nxt = sequence[i + 1]
    if curr in nucs and nxt in nucs:
        count_matrix[curr][nxt] += 1
```

## Normalize to Get Transition Probabilities
We convert the counts to probabilities by dividing each row by its total.

```python
transition_matrix = {}
for curr in nucs:
    total = sum(count_matrix[curr].values())
    if total == 0:
        transition_matrix[curr] = {n: 0 for n in nucs}
    else:
        transition_matrix[curr] = {n: count_matrix[curr][n] / total for n in nucs}
```

## Transition Matrix

```python
import pandas as pd
df = pd.DataFrame(transition_matrix).T
print("Markov Transition Matrix (Probabilities):")
print(df.round(3))
```
