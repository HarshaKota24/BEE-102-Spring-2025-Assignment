## Convert Multi-line FASTA to Single-line FASTA

```python
input_file =  "multiline_input.fasta"  
output_file = "singleline_output.fasta"

with open(input_file, "r") as infile, open(output_file, "w") as outfile:
    seq_lines = []
    for line in infile:
        line = line.strip()
        if line.startswith(">"):
            if seq_lines:
                outfile.write("".join(seq_lines) + "\n")
                seq_lines = []
            outfile.write(line + "\n")
        else:
            seq_lines.append(line)
    if seq_lines:
        outfile.write("".join(seq_lines) + "\n")
## Results are stored in "singleline_output.fasta"
```
